## Molecules from SMILES

```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Draw

mol = Chem.MolFromSmiles("CC(=O)Oc1ccccc1C(=O)O")  # aspirin
mol
```

```python
print(f"MW   = {Descriptors.MolWt(mol):.1f} g/mol")
print(f"logP = {Descriptors.MolLogP(mol):.2f}")
print(f"TPSA = {Descriptors.TPSA(mol):.1f} Å²")
```

## Several molecules at once

```python
smiles = {
    "aspirin": "CC(=O)Oc1ccccc1C(=O)O",
    "ibuprofen": "CC(C)Cc1ccc(cc1)C(C)C(=O)O",
    "caffeine": "CN1C=NC2=C1C(=O)N(C(=O)N2C)C",
    "paracetamol": "CC(=O)Nc1ccc(O)cc1",
}
mols = [Chem.MolFromSmiles(s) for s in smiles.values()]
Draw.MolsToGridImage(mols, legends=list(smiles), molsPerRow=4)
```

## A 3D structure

We add hydrogens, embed the molecule in 3D and do a quick force-field (MMFF) optimisation.

```python
from rdkit.Chem import AllChem
import py3Dmol

mol3d = Chem.AddHs(mol)
AllChem.EmbedMolecule(mol3d, randomSeed=42)
AllChem.MMFFOptimizeMolecule(mol3d)

view = py3Dmol.view(width=400, height=300)
view.addModel(Chem.MolToMolBlock(mol3d), "mol")
view.setStyle({"stick": {}})
view.zoomTo()
view.show()
```
