## Structure

```python
from io import StringIO
from ase.io import read

xyz = """2
N2
N   0.000   0.000   0.000
N   0.000   0.000   1.100
"""

atoms = read(StringIO(xyz), format="xyz")
print(f"N-N distance: {atoms.get_distance(0, 1):.3f} Å")
```

## Single point

```python
from pyscf import gto, scf

BASIS = "def2-svp"


def single_point(atoms):
    """HF energy (Hartree) of an ASE Atoms object."""
    mol = gto.M(
        atom=[(s, tuple(p)) for s, p in zip(atoms.get_chemical_symbols(), atoms.positions)],
        basis=BASIS,
        unit="Angstrom",
        verbose=0,
    )
    return scf.RHF(mol).run().e_tot


energy = single_point(atoms)
print(f"E = {energy:.6f} Hartree")
```

## Scan

```python
import numpy as np

distances = np.linspace(0.90, 1.60, 15)
energies = []

scan_atoms = atoms.copy()
for d in distances:
    scan_atoms.set_distance(0, 1, d)
    e = single_point(scan_atoms)
    energies.append(e)
    print(f"{d:.3f} Å   {e:.6f} Hartree")
```

```python
import matplotlib.pyplot as plt

plt.plot(distances, energies, "o-")
plt.xlabel("N–N distance (Å)")
plt.ylabel("Energy (Hartree)")
plt.show()
```
