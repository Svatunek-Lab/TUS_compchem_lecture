## Building a molecule

```python
from ase.build import molecule

ethanol = molecule("CH3CH2OH")
print(ethanol.get_chemical_formula())
for i, (symbol, pos) in enumerate(zip(ethanol.get_chemical_symbols(), ethanol.positions)):
    print(f"{i:2d} {symbol:2s} {pos[0]:8.3f} {pos[1]:8.3f} {pos[2]:8.3f}")
```

## Measuring distances and angles

Atoms are addressed by their index (see the list above).

```python
print(f"C-C distance: {ethanol.get_distance(0, 1):.3f} Å")
print(f"C-O distance: {ethanol.get_distance(1, 2):.3f} Å")
print(f"C-C-O angle:  {ethanol.get_angle(0, 1, 2):.1f}°")
```

## Writing, reading and viewing files

```python
from ase.io import read, write
import py3Dmol

write("ethanol.xyz", ethanol)
print(open("ethanol.xyz").read())

view = py3Dmol.view(width=400, height=300)
view.addModel(open("ethanol.xyz").read(), "xyz")
view.setStyle({"stick": {}})
view.zoomTo()
view.show()
```

## From ASE to PySCF

PySCF takes the geometry as a string of `symbol x y z` lines. We will use this in the next notebook.

```python
def atoms_to_pyscf(atoms):
    return "\n".join(
        f"{s} {x:.6f} {y:.6f} {z:.6f}"
        for s, (x, y, z) in zip(atoms.get_chemical_symbols(), atoms.positions)
    )

print(atoms_to_pyscf(ethanol))
```
