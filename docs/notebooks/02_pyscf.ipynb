## Define the molecule

```python
from pyscf import gto, scf, dft

mol = gto.M(
    atom="""
    O  0.000  0.000  0.117
    H  0.000  0.757 -0.467
    H  0.000 -0.757 -0.467
    """,
    basis="def2-svp",
)
print(f"{mol.nelectron} electrons, {mol.nao} basis functions")
```

## Hartree–Fock and DFT

```python
mf_hf = scf.RHF(mol).run()
mf_dft = dft.RKS(mol, xc="b3lyp").run()

print(f"HF    energy: {mf_hf.e_tot:.6f} Eh")
print(f"B3LYP energy: {mf_dft.e_tot:.6f} Eh")
```

## Orbital energies and dipole moment

```python
HARTREE_TO_EV = 27.211386

n_occ = mol.nelectron // 2
homo = mf_dft.mo_energy[n_occ - 1] * HARTREE_TO_EV
lumo = mf_dft.mo_energy[n_occ] * HARTREE_TO_EV
print(f"HOMO = {homo:.2f} eV, LUMO = {lumo:.2f} eV, gap = {lumo - homo:.2f} eV")

mf_dft.dip_moment()
```

## Geometry optimisation

```python
from pyscf.geomopt.geometric_solver import optimize

mol_opt = optimize(mf_dft)
print(mol_opt.tostring(format="xyz"))
```
